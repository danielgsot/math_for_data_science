# Vectors for Data Science

This notebook reviews the fundamental concepts of vectors used in Data Science and Machine Learning.

Topics covered:
- Vector representation
- Vector addition and subtraction
- Scalar multiplication
- Dot product
- Vector norm
- Distance between vectors
- Angle between vectors
- Linear combinations

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. What is a vector?

A vector is an ordered collection of numbers.

For example:

$$
\mathbf{x} =
\begin{bmatrix}
2 \\
3
\end{bmatrix}
$$

In Data Science, a vector can represent:

- one data sample
- a set of features
- model parameters
- gradients
- embeddings

For example, a person could be represented using three features:

$$
\mathbf{x} =
\begin{bmatrix}
22 \\
1.80 \\
75
\end{bmatrix}
$$

where:

- $x_1$ = age
- $x_2$ = height
- $x_3$ = weight

This is called a **feature vector**.

Since it contains three components:

$$
\mathbf{x} \in \mathbb{R}^3
$$

```python
person = np.array([22, 1.80, 75])

print(person)
print("Shape:", person.shape)
print("Data type:", person.dtype)
```

## 2. Creating vectors with NumPy

NumPy represents vectors using one-dimensional arrays.

For example:

$$
\mathbf{x} =
\begin{bmatrix}
2 \\
3
\end{bmatrix}
$$

```python
x = np.array([2, 3])

print(x)
print("Shape:", x.shape)
print("Dimensions:", x.ndim)
```

## 3. Visual representation

A vector has both **direction** and **magnitude**.

For example:

$$
\mathbf{x} =
\begin{bmatrix}
2 \\
3
\end{bmatrix}
$$

can be represented geometrically as an arrow from the origin $(0,0)$ to the point $(2,3)$.

```python
x = np.array([2, 3])

plt.figure(figsize=(6, 6))

plt.quiver(
    0,
    0,
    x[0],
    x[1],
    angles="xy",
    scale_units="xy",
    scale=1
)

plt.xlim(-1, 4)
plt.ylim(-1, 4)

plt.axhline(0)
plt.axvline(0)

plt.grid()

plt.xlabel("x")
plt.ylabel("y")

plt.title("Vector x = [2, 3]")

plt.show()
```

## 4. Vector addition

Two vectors of the same dimension can be added component by component.

Consider:

$$
\mathbf{x} =
\begin{bmatrix}
2 \\
3
\end{bmatrix}
$$

and:

$$
\mathbf{y} =
\begin{bmatrix}
1 \\
4
\end{bmatrix}
$$

Then:

$$
\mathbf{x} + \mathbf{y}
=
\begin{bmatrix}
2 + 1 \\
3 + 4
\end{bmatrix}
=
\begin{bmatrix}
3 \\
7
\end{bmatrix}
$$

```python
x = np.array([2, 3])
y = np.array([1, 4])

z = x + y

print("x =", x)
print("y =", y)
print("x + y =", z)
```

## 5. Vector subtraction

Vector subtraction is also performed component by component.

$$
\mathbf{x} - \mathbf{y}
=
\begin{bmatrix}
x_1 - y_1 \\
x_2 - y_2
\end{bmatrix}
$$

Geometrically, $\mathbf{x} - \mathbf{y}$ can be interpreted as the vector pointing from $\mathbf{y}$ to $\mathbf{x}$.

```python
x = np.array([5, 8])
y = np.array([2, 3])

result = x - y

print("x =", x)
print("y =", y)
print("x - y =", result)
```

## 6. Scalar multiplication

A vector can be multiplied by a scalar.

If:

$$
\mathbf{x} =
\begin{bmatrix}
2 \\
3
\end{bmatrix}
$$

then:

$$
2\mathbf{x}
=
\begin{bmatrix}
4 \\
6
\end{bmatrix}
$$

Multiplying a vector by a scalar changes its magnitude.

A negative scalar also reverses its direction.

```python
x = np.array([2, 3])

print("x =", x)
print("2x =", 2 * x)
print("0.5x =", 0.5 * x)
print("-x =", -1 * x)
```

## 7. Vector norm

The norm of a vector measures its magnitude or length.

The Euclidean norm is defined as:

$$
\|\mathbf{x}\|_2
=
\sqrt{x_1^2 + x_2^2 + \cdots + x_n^2}
$$

For example:

$$
\mathbf{x} =
\begin{bmatrix}
3 \\
4
\end{bmatrix}
$$

Then:

$$
\|\mathbf{x}\|_2
=
\sqrt{3^2 + 4^2}
=
5
$$

```python
x = np.array([3, 4])

norm = np.linalg.norm(x)

print("Vector:", x)
print("Norm:", norm)
```

```python
manual_norm = np.sqrt(np.sum(x ** 2))

print("Manual norm:", manual_norm)
```

The norm is important in Machine Learning because it is used in areas such as:

- distance calculations
- regularization
- optimization
- normalization
- similarity measures

## 8. Dot product

The dot product is one of the most important vector operations in Machine Learning.

For two vectors:

$$
\mathbf{x} =
\begin{bmatrix}
x_1 \\
x_2
\end{bmatrix}
$$

and:

$$
\mathbf{y} =
\begin{bmatrix}
y_1 \\
y_2
\end{bmatrix}
$$

their dot product is:

$$
\mathbf{x}^T\mathbf{y}
=
x_1y_1 + x_2y_2
$$

The result of a dot product is a scalar.

```python
x = np.array([1, 2])
y = np.array([3, 4])

dot_product = np.dot(x, y)

print("x =", x)
print("y =", y)
print("Dot product:", dot_product)
```

For these vectors:

$$
\mathbf{x}^T\mathbf{y}
=
1 \cdot 3 + 2 \cdot 4
$$

Therefore:

$$
\mathbf{x}^T\mathbf{y}
=
3 + 8
=
11
$$

```python
dot_product = x @ y

print(dot_product)
```

## 9. Distance between vectors

The Euclidean distance between two vectors is calculated using:

$$
d(\mathbf{x}, \mathbf{y})
=
\|\mathbf{x} - \mathbf{y}\|_2
$$

This concept is particularly important in algorithms such as:

- K-Nearest Neighbors
- K-Means
- clustering
- similarity search

```python
x = np.array([1, 2])
y = np.array([4, 6])

distance = np.linalg.norm(x - y)

print("x =", x)
print("y =", y)
print("Euclidean distance:", distance)
```

## 10. Angle between vectors

The dot product is related to the angle between two vectors:

$$
\mathbf{x} \cdot \mathbf{y}
=
\|\mathbf{x}\|
\|\mathbf{y}\|
\cos(\theta)
$$

Therefore:

$$
\cos(\theta)
=
\frac{
\mathbf{x} \cdot \mathbf{y}
}{
\|\mathbf{x}\|
\|\mathbf{y}\|
}
$$

And the angle can be calculated using:

$$
\theta =
\arccos
\left(
\frac{
\mathbf{x} \cdot \mathbf{y}
}{
\|\mathbf{x}\|
\|\mathbf{y}\|
}
\right)
$$

```python
x = np.array([1, 0])
y = np.array([1, 1])

cos_theta = np.dot(x, y) / (
    np.linalg.norm(x) * np.linalg.norm(y)
)

theta_radians = np.arccos(cos_theta)
theta_degrees = np.degrees(theta_radians)

print("cos(theta):", cos_theta)
print("Angle in radians:", theta_radians)
print("Angle in degrees:", theta_degrees)
```

## 11. Orthogonal vectors

Two vectors are orthogonal when the angle between them is $90^\circ$.

This happens when their dot product is zero:

$$
\mathbf{x} \cdot \mathbf{y} = 0
$$

For example:

$$
\mathbf{x} =
\begin{bmatrix}
1 \\
0
\end{bmatrix}
$$

and:

$$
\mathbf{y} =
\begin{bmatrix}
0 \\
1
\end{bmatrix}
$$

are orthogonal.

```python
x = np.array([1, 0])
y = np.array([0, 1])

dot_product = np.dot(x, y)

print("Dot product:", dot_product)
```

## 12. Linear combinations

A linear combination is obtained by multiplying vectors by scalars and adding the results.

Given:

$$
\mathbf{v}_1 =
\begin{bmatrix}
1 \\
0
\end{bmatrix}
$$

and:

$$
\mathbf{v}_2 =
\begin{bmatrix}
0 \\
1
\end{bmatrix}
$$

we can construct:

$$
\mathbf{x}
=
3\mathbf{v}_1
+
2\mathbf{v}_2
$$

Therefore:

$$
\mathbf{x}
=
\begin{bmatrix}
3 \\
2
\end{bmatrix}
$$

Linear combinations are fundamental for understanding concepts such as:

- vector spaces
- span
- basis
- linear independence
- matrix multiplication
- Principal Component Analysis (PCA)

## 13. Vectors in Machine Learning

Vectors appear everywhere in Machine Learning.

A data sample with $n$ features can be represented as:

$$
\mathbf{x}
=
\begin{bmatrix}
x_1 \\
x_2 \\
\vdots \\
x_n
\end{bmatrix}
$$

A linear model can be written as:

$$
\hat{y}
=
\mathbf{w}^T\mathbf{x}
+
b
$$

where:

- $\mathbf{x}$ is the feature vector
- $\mathbf{w}$ is the vector of model parameters
- $b$ is the bias
- $\hat{y}$ is the prediction

```python
x = np.array([22, 1.80, 75])

w = np.array([0.5, 3, 0.1])

b = 2

prediction = w @ x + b

print("Feature vector:", x)
print("Weights:", w)
print("Bias:", b)
print("Prediction:", prediction)
```

The expression:

$$
\mathbf{w}^T\mathbf{x}
$$

is simply a dot product.

For this reason, understanding vectors and dot products is fundamental before studying models such as Linear Regression and Logistic Regression.
